```python
import nltk
from nltk.corpus import stopwords 
from nltk.stem import WordNetLemmatizer
import re 
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
    # handling missing/non string values
    if not isinstance(text,str):
        return ""
    # lowercase
    # remove puncuation
    text = str(text).lower()
    text = re.sub(r'[^a-zA-Z\s]','',text)
    # tokenization and remove sstopwords
    words = text.split()
    words = [word for word in words if word not in stop_words]
    # lemmatization 
    words = [lemmatizer.lemmatize(word) for word in words]
    # list to string 

    return " ".join(words)
```

```python
df['tags'] = df['tags'].apply(preprocess_text)
```

```python
df.head()
```

```python
# converting each sentences into vector
```

```python
df = df.reset_index(drop=True)
# koi indexmis missing then remve it and reindex all
```

```python
indices = pd.Series(df.index,index=df['title'])

indices = indices[~indices.index.duplicated(keep='first')]
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
# dict soize 500k
```

```python
# conver to vector if all then it would be 2D = matrix
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix
```

```python
# Cosine Similarity 
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n=10):
    if title not in indices:
        return ['Movie not found']

    idx = indices[title]

    print("Index:", idx)
    print("Movie:", df.iloc[idx]['title'])
    print("Tags:", df.iloc[idx]['tags'])
    print("Number of TF-IDF words:", tfidf_matrix[idx].nnz)

    sim_score = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    print("Maximum similarity:", sim_score.max())
    print("Non-zero similarities:", np.count_nonzero(sim_score))

    similar_idx = sim_score.argsort()[::-1][1:n+1]

    return df['title'].iloc[similar_idx]
```

```python
recommend('Avatar')
```

```python
import pickle
pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))

pickle.dump(indices,open('indices.pkl','wb'))
df.to_pickle('df.pkl')
pickle.dump(tfidf,open('tfidf.pkl','wb'))
```
